# 00 â€” Controls and Baselines (run this FIRST)

Every representation-similarity number in notebooks 01â€“05 needs an answer to two questions before it means anything:

1. **What does a model with *no* learned Tamil knowledge score?** (random-init / chance controls)
2. **What does plain word-overlap already achieve?** (TF-IDF lexical baseline â€” no neural network at all)

This notebook computes both on the corrected corpus (**1,262 pairs, 393 Naaladiyar**), plus a demonstration of
**why the raw CCA r â‰ˆ 1.000 numbers from the v1 LSTM study are degenerate** and how to compute a non-degenerate
(PCA-reduced) CCA instead.

Needs only the Python standard library + numpy; the last section optionally uses scikit-learn if present.

```python
import json, math, re, random
from collections import Counter
from pathlib import Path

SEED = 3407

def normalize_text(text):
    text = str(text).replace("\ufeff", " ").replace("\r\n", "\n")
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()

def tokenize(text):
    text = normalize_text(text).lower().replace("\n", " ")
    toks = []
    for raw in text.split():
        cleaned = re.sub(r"[^\w\u0B80-\u0BFF]+", "", raw)
        if cleaned:
            toks.append(cleaned)
    return toks

FILES = {
    "naladiyar": "naladiyar.jsonl",
    "tholkappiyam_eth": "sft_data-eth.jsonl",
    "tholkappiyam_sol": "sft_data-sol.jsonl",
    "tholkappiyam_por": "sft_data-por.jsonl",
    "thirukadukam": "thirukadukam_tamilvu.jsonl",
}
DATA_DIR = Path("data")

rows = []
for ds, fn in FILES.items():
    with open(DATA_DIR / fn, encoding="utf-8-sig", errors="ignore") as f:
        for line in f:
            if not line.strip():
                continue
            obj = json.loads(line)
            v = normalize_text(obj.get("verse", ""))
            u = normalize_text(obj.get("explanation", obj.get("explaination", obj.get("urai", ""))))
            if v and u:
                rows.append({"dataset": ds, "verse": v, "urai": u})

c = Counter(r["dataset"] for r in rows)
print("per-dataset:", dict(c), " total:", len(rows))
assert len(rows) == 1262, f"expected 1,262 rows, got {len(rows)}"
assert c["naladiyar"] == 393, f"expected 393 Naaladiyar, got {c['naladiyar']}"
print("OK: corrected corpus confirmed")
```

## Control 1 â€” TF-IDF lexical retrieval baseline

For every verse, rank **all** urais in the same pool by TF-IDF cosine similarity and check where the true urai lands.
No neural network, no training â€” this is what pure word overlap buys. Any learned representation that cannot beat
these numbers has learned nothing beyond surface vocabulary.

Implemented with dictionaries (no sklearn needed), sublinear tf, smoothed idf, L2-normalised.

```python
def tfidf_vecs(docs_tokens):
    df = Counter()
    for toks in docs_tokens:
        df.update(set(toks))
    N = len(docs_tokens)
    idf = {t: math.log((N + 1) / (cnt + 1)) + 1 for t, cnt in df.items()}
    vecs = []
    for toks in docs_tokens:
        tf = Counter(toks)
        vec = {t: (1 + math.log(cnt)) * idf[t] for t, cnt in tf.items()}
        norm = math.sqrt(sum(w * w for w in vec.values())) or 1.0
        vecs.append({t: w / norm for t, w in vec.items()})
    return vecs

def cos(a, b):
    if len(b) < len(a):
        a, b = b, a
    return sum(w * b.get(t, 0.0) for t, w in a.items())

def retrieval_eval(subrows, label):
    verses = [tokenize(r["verse"]) for r in subrows]
    urais  = [tokenize(r["urai"])  for r in subrows]
    all_vecs = tfidf_vecs(verses + urais)
    vv, uv = all_vecs[:len(verses)], all_vecs[len(verses):]
    n = len(subrows)
    r1 = r5 = r10 = 0
    mrr = 0.0
    cos_matched, cos_mismatched = [], []
    rng = random.Random(SEED)
    for i in range(n):
        sims = sorted(((cos(vv[i], uv[j]), j) for j in range(n)), reverse=True)
        rank = next(k for k, (_, j) in enumerate(sims) if j == i) + 1
        r1 += rank <= 1; r5 += rank <= 5; r10 += rank <= 10
        mrr += 1.0 / rank
        cos_matched.append(cos(vv[i], uv[i]))
        j = rng.randrange(n)
        cos_mismatched.append(cos(vv[i], uv[(j + 1) % n if j == i else j]))
    mean = lambda x: sum(x) / len(x)
    print(f"[{label:18s}] n={n:5d}  R@1={r1/n:.3f}  R@5={r5/n:.3f}  R@10={r10/n:.3f}  MRR={mrr/n:.3f}  "
          f"cos(matched)={mean(cos_matched):.3f}  cos(random)={mean(cos_mismatched):.3f}  chance R@1={1/n:.4f}")

for ds in FILES:
    retrieval_eval([r for r in rows if r["dataset"] == ds], ds)
retrieval_eval([r for r in rows if r["dataset"].startswith("tholkappiyam")], "tholkappiyam_all")
retrieval_eval(rows, "combined")
```

## Control 2 â€” Why the v1 CCA r â‰ˆ 1.000 numbers are degenerate

Classical CCA finds projections maximising correlation between two views. When the number of samples *n* is
close to (or below) the representation dimension *d*, CCA can achieve **perfect correlation on any data,
including pure noise**, because the optimisation has more free parameters than constraints.

v1 used d = 256-dim sentence vectors with n = 393 (Naaladiyar) and n = 100 (Thirukadukam) â€” the r â‰ˆ 1.000
values there are an artefact, not evidence of alignment. The cell below proves it with random Gaussian
vectors, then shows the partial fix: **PCA-reduce both views (e.g. to 32 dims) before CCA**. As the executed
output below shows, this *reduces* the bias but does not eliminate it at these sample sizes (random data
still reaches r1 = 0.905 at n=100 and 0.542 at n=393 with k=32). The only valid protocol is therefore:
report trained-model CCA **next to the random-data control at identical n and k** and interpret only the
gap between them; for the n=100 pool, also use a stronger reduction (k=8 or 16).

```python
import numpy as np

rng = np.random.default_rng(SEED)

def simple_cca_first_corr(X, Y, reg=1e-8):
    # first canonical correlation via the standard generalised-eigenvalue route
    Xc = X - X.mean(0); Yc = Y - Y.mean(0)
    n = X.shape[0]
    Sxx = Xc.T @ Xc / n + reg * np.eye(X.shape[1])
    Syy = Yc.T @ Yc / n + reg * np.eye(Y.shape[1])
    Sxy = Xc.T @ Yc / n
    A = np.linalg.solve(Sxx, Sxy)
    B = np.linalg.solve(Syy, Sxy.T)
    eigvals = np.linalg.eigvals(A @ B)
    return float(np.sqrt(np.max(np.real(eigvals)).clip(0, 1)))

def pca_reduce(X, k):
    Xc = X - X.mean(0)
    U, S, Vt = np.linalg.svd(Xc, full_matrices=False)
    return Xc @ Vt[:k].T

for n in (100, 393):
    X = rng.standard_normal((n, 256))
    Y = rng.standard_normal((n, 256))     # completely independent noise
    r_raw = simple_cca_first_corr(X, Y)
    Xr, Yr = pca_reduce(X, 32), pca_reduce(Y, 32)
    r_pca = simple_cca_first_corr(Xr, Yr)
    print(f"n={n:4d}, d=256 | raw CCA r1 on RANDOM data = {r_raw:.3f}   "
          f"| PCA-32 CCA r1 on RANDOM data = {r_pca:.3f}")

print()
print("Conclusion: any raw-CCA value computed at n <= d is uninterpretable.")
print("Recipe for notebooks 01/02: PCA both views to 32 dims, then CCA, and")
print("report the PCA-CCA number next to a random-vector control at the same n.")
```
