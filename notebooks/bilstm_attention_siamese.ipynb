# Three Ways to Read Tamil: BiLSTM, Attention, and Siamese Matching

This notebook tries **three different encoder designs** on the same verse/commentary data, to see which one actually learns to connect a verse to its own explanation, rather than just learning "the vocabulary of Tamil in general."

**Part A â€” BiLSTM.** Two separate bidirectional-LSTM models per text: one that only reads verses, one that only reads commentary (urai). Compares individual *words* across the two models by cosine similarity.

**Part B â€” Attention Encoders.** Same two-separate-models idea, but each side is a small Transformer (self-attention) encoder instead of an LSTM.

**Part C â€” Siamese Attention Networks.** Takes the Part B attention encoders and adds a training objective that *directly* teaches the model to recognise real verseâ€“urai pairs vs. random mismatched ones.

**8 models each for Part A and Part B** (Naaladiyar / Thirukadukam / Tholkappiyam / combined Ã— verse+urai), **4 Siamese networks for Part C** (one per nool + combined). No train/test split â€” this is representation *analysis*, not a benchmark.

**Headline result to watch for:** Part A's raw word-similarity matching finds pairs that *look* similar in embedding space, but almost none of them (0 or 1 out of the top 20, every single time) actually come from the same real verseâ€“urai example â€” see Part A below. Part C's Siamese network, which is *trained specifically* to spot real pairs, does far better â€” but even it shows the same "shortcut" pattern we saw elsewhere: separation collapses once all texts are combined, suggesting it's partly learning "which text is this" rather than genuine verseâ€“urai matching.

```python
def normalize_text(text: str) -> str:
    text = str(text).replace("\ufeff", " ").replace("\r\n", "\n")
    text = re.sub(r"[ \t]+", " ", text)
    return text.strip()

# --- v3: minimal special-token set -------------------------------------------
# Removed vs v2, deliberately:
#   * dataset identity tags <naladiyar>/<thirukadukam>/<tholkappiyam>
#     -> they let a model win by recognising WHICH text a pair came from
#        (the source-identification confound behind the 0.78 combined
#        pair-classifier) instead of whether verse and urai actually match.
#   * <lb> line-break token
#     -> tokenize_words only emitted it for text that still had newlines. In
#        the grammar probe the ORIGINAL verse kept its newlines while every
#        scrambled variant was rebuilt as a single line, so the original was
#        scored with (n_lines - 1) extra, highly predictable <lb> tokens that
#        no competitor had. That inflated the 19/19 result. Gone.
PAD, UNK, MASK, BOS, EOS = "<pad>", "<unk>", "<mask>", "<bos>", "<eos>"
VERSE_TAG, URAI_TAG = "<verse>", "<urai>"

SPECIAL_TOKENS = [PAD, UNK, MASK, BOS, EOS, VERSE_TAG, URAI_TAG]

DS_COLORS = {
    "naladiyar":        "tab:blue",
    "thirukadukam":     "tab:purple",
    "tholkappiyam_eth": "tab:orange",
    "tholkappiyam_por": "tab:green",
    "tholkappiyam_sol": "tab:red",
}

def dataset_name_from_path(path):
    p = str(path).lower()
    if "thirukadukam" in p: return "thirukadukam"
    if "naladiyar" in p:    return "naladiyar"
    if "eth" in p:          return "tholkappiyam_eth"
    if "por" in p:          return "tholkappiyam_por"
    if "sol" in p:          return "tholkappiyam_sol"
    return Path(path).stem

def tokenize_words(text: str):
    """v3: flat word list. Line structure is NOT encoded (see the <lb> note
    above), so a verse and any re-linearised rewrite of it are tokenized on
    exactly equal terms."""
    text = normalize_text(text).lower()
    tokens = []
    for word in text.split():
        cleaned = re.sub(r"[^\w\u0B80-\u0BFF]+", "", word)
        if cleaned: tokens.append(cleaned)
    return tokens

def encode_tokens(tokens, stoi):
    unk = stoi.get(UNK, 1)
    return [stoi.get(t, unk) for t in tokens]
```

```python
DATA_PATHS = [
    "data/naladiyar.jsonl",
    "data/sft_data-eth.jsonl",
    "data/sft_data-por.jsonl",
    "data/sft_data-sol.jsonl",
    "data/thirukadukam_tamilvu.jsonl",
]

def load_rows(paths):
    rows = []
    for path in paths:
        ds  = dataset_name_from_path(path)
        enc = "utf-8-sig" if "naladiyar" in str(path).lower() else "utf-8"
        with open(path, "r", encoding=enc, errors="ignore") as f:
            for line in f:
                if not line.strip(): continue
                obj   = json.loads(line)
                verse = normalize_text(obj.get("verse", ""))
                urai  = normalize_text(obj.get("explanation", obj.get("explaination", obj.get("urai", ""))))
                if verse and urai:
                    rows.append({"dataset": ds, "verse": verse, "urai": urai})
    return rows

all_rows = load_rows(DATA_PATHS)
df_all   = pd.DataFrame(all_rows)
print(f"Total rows: {len(all_rows)}")
print(df_all["dataset"].value_counts())
```

```python
def build_vocab(rows, min_freq=1):
    counts = Counter()
    for row in rows:
        counts.update(tokenize_words(row["verse"]))
        counts.update(tokenize_words(row["urai"]))
    stoi = {tok: i for i, tok in enumerate(SPECIAL_TOKENS)}
    for tok, freq in counts.items():
        if freq >= min_freq and tok not in stoi:
            stoi[tok] = len(stoi)
    return stoi

shared_vocab = build_vocab(all_rows)
shared_itos  = {v: k for k, v in shared_vocab.items()}
print(f"Vocab size: {len(shared_vocab)}")
```

```python
def get_nool_rows(nool, rows):
    if nool == "tholkappiyam":
        return [r for r in rows if r["dataset"].startswith("tholkappiyam")]
    if nool == "combined":
        return rows
    return [r for r in rows if r["dataset"] == nool]

NOOLS = ["naladiyar", "thirukadukam", "tholkappiyam"]
```

```python
def mask_tokens(ids, mask_id, mask_prob=0.15):
    x = ids.copy(); y = [-100] * len(ids)
    cands = list(range(1, len(ids)))
    if not cands: return x, y
    forced = random.choice(cands)
    for i in cands:
        if random.random() < mask_prob or i == forced:
            y[i] = ids[i]; x[i] = mask_id
    return x, y

class MLMDataset(Dataset):
    def __init__(self, texts, vocab, prefix_toks):
        self.items = []
        for text in texts:
            toks = prefix_toks + tokenize_words(text)
            ids  = encode_tokens(toks, vocab)
            if len(ids) >= 3: self.items.append(ids)
    def __len__(self): return len(self.items)
    def __getitem__(self, i): return self.items[i]

def mlm_collate(batch, vocab):
    pad_id  = vocab[PAD]; mask_id = vocab[MASK]
    inputs, labels, lengths = [], [], []
    for ids in batch:
        x, y = mask_tokens(ids, mask_id)
        inputs.append(x); labels.append(y); lengths.append(len(x))
    mlen = max(lengths)
    inp_t = torch.full((len(batch), mlen), pad_id, dtype=torch.long)
    lab_t = torch.full((len(batch), mlen), -100,   dtype=torch.long)
    msk_t = torch.zeros(len(batch), mlen, dtype=torch.bool)
    for i, (x, y, l) in enumerate(zip(inputs, labels, lengths)):
        inp_t[i, :l] = torch.tensor(x)
        lab_t[i, :l] = torch.tensor(y)
        msk_t[i, :l] = True
    return inp_t, lab_t, msk_t, torch.tensor(lengths, dtype=torch.long)
```

## Part A â€” BiLSTM (separate verse + urai models)

Two independent `BiLSTMEncoder`s trained per text with masked-language-modelling (guess the blanked-out word) â€” one only ever sees verses, one only ever sees commentary. Neither model is told anything about the other. 20 epochs, hidden size 192Ã—2 (bidirectional), 2 layers.

**Real result from this run** â€” masked-word prediction loss, lower is better:

| | Naaladiyar | Thirukadukam | Tholkappiyam | Combined |
|---|---|---|---|---|
| verse | 6.289 | 6.566 | 4.725 | 5.361 |
| urai | 6.873 | 7.690 | 6.697 | 6.388 |

Same pattern as every other encoder in this project: urai (commentary) is consistently harder to predict than the terse verse text, because it's longer, freer prose.

```python
class BiLSTMEncoder(nn.Module):
    """BiLSTM encoder trained separately on verse OR urai sequences.
    Returns (lm_logits, per_token_hidden_states).
    hidden_states shape: [batch, seq_len, hidden_dim*2]
    """
    def __init__(self, vocab_size, emb_dim=128, hidden_dim=192,
                 num_layers=2, dropout=0.15, pad_id=0):
        super().__init__()
        self.out_dim = hidden_dim * 2
        self.embedding = nn.Embedding(vocab_size, emb_dim, padding_idx=pad_id)
        self.lstm = nn.LSTM(emb_dim, hidden_dim, num_layers=num_layers,
                            batch_first=True, bidirectional=True,
                            dropout=dropout if num_layers > 1 else 0.0)
        self.lm_head = nn.Linear(hidden_dim * 2, vocab_size)

    def forward(self, ids, lengths):
        x      = self.embedding(ids)
        packed = nn.utils.rnn.pack_padded_sequence(
                     x, lengths.cpu(), batch_first=True, enforce_sorted=False)
        out, _ = self.lstm(packed)
        out, _ = nn.utils.rnn.pad_packed_sequence(out, batch_first=True)
        return self.lm_head(out), out
```

```python
def train_bilstm(model, texts, vocab, prefix_toks, name, epochs=20, bs=32, lr=2e-3):
    ds     = MLMDataset(texts, vocab, prefix_toks)
    loader = DataLoader(ds, batch_size=bs, shuffle=True,
                        collate_fn=lambda b: mlm_collate(b, vocab))
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-2)
    model.to(DEVICE); hist = []
    for ep in range(epochs):
        model.train(); total = n = 0
        for inp, lab, msk, lengths in loader:
            inp = inp.to(DEVICE); lab = lab.to(DEVICE); lengths = lengths.to(DEVICE)
            opt.zero_grad()
            logits, _ = model(inp, lengths)
            loss = F.cross_entropy(logits.reshape(-1, logits.size(-1)),
                                   lab.reshape(-1), ignore_index=-100)
            loss.backward()
            nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            opt.step()
            total += loss.item() * inp.size(0); n += inp.size(0)
        hist.append(total / n)
    print(f"  [BiLSTM {name:30s}]  final_loss={hist[-1]:.4f}")
    model.cpu(); return hist

# Two separate BiLSTM models per nool: verse encoder + urai encoder
bilstm_verse = {}; bilstm_urai = {}; bilstm_hist = {}
print("=== Training BiLSTM encoders (separate verse + urai, per nool + combined) ===")
for nool in NOOLS + ["combined"]:
    nool_rows = get_nool_rows(nool, all_rows)
    verses    = [r["verse"] for r in nool_rows]
    urais     = [r["urai"]  for r in nool_rows]
    print(f"\n[{nool}]  n={len(nool_rows)}")

    mv = BiLSTMEncoder(len(shared_vocab), pad_id=shared_vocab[PAD])
    mu = BiLSTMEncoder(len(shared_vocab), pad_id=shared_vocab[PAD])
    hv = train_bilstm(mv, verses, shared_vocab, [VERSE_TAG], f"{nool}/verse", epochs=20)
    hu = train_bilstm(mu, urais,  shared_vocab, [URAI_TAG],  f"{nool}/urai",  epochs=20)
    bilstm_verse[nool] = mv; bilstm_urai[nool] = mu
    bilstm_hist[nool]  = (hv, hu)

fig, axes = plt.subplots(1, len(NOOLS)+1, figsize=(4*(len(NOOLS)+1), 3), sharey=True)
for ax, nool in zip(axes, NOOLS + ["combined"]):
    hv, hu = bilstm_hist[nool]
    ax.plot(hv, label="verse"); ax.plot(hu, label="urai")
    ax.set_title(nool, fontsize=9); ax.set_xlabel("Epoch")
    ax.grid(True, alpha=0.3); ax.legend(fontsize=7)
axes[0].set_ylabel("Masked-LM loss")
plt.suptitle("BiLSTM Training Curves (separate verse / urai models)", fontsize=11)
plt.tight_layout(); plt.show()
```

## BiLSTM word-level cosine similarity

Since the verse-model and urai-model were trained completely separately, we ask: if we average every occurrence of each word into one vector, do any *verse words* end up pointing in a similar direction to any *urai words*? Full verse-vocabulary Ã— urai-vocabulary cosine similarity is computed, and the **top-20** cross-modal word pairs are kept (each word used at most once, so it's genuinely 20 distinct verse words paired with 20 distinct urai words).

Each pair is also checked against the real training data: does this verse word and urai word ever actually appear **together in the same real verseâ€“urai example** (`same_verse`), or are they only similar by coincidence across two *different* examples (`cross_verse`)?

**Real result â€” this is the headline finding of Part A:** across every single nool, the overall similarity is barely above zero (Naaladiyar mean **-0.011**, Thirukadukam **0.018**, Tholkappiyam **-0.020**, combined **0.003** â€” essentially noise), and of the top-20 highest-scoring word pairs in each case, **0 out of 20 are same-verse pairs** for Naaladiyar/Tholkappiyam/combined, and only **1 out of 20** for Thirukadukam.

**What this means in plain terms:** the "most similar" verse-word/urai-word matches this method finds are almost never real matches â€” they're words that just happen to land near each other in embedding space by chance, having never appeared in the same verse-and-commentary pair at all. Two separately-trained BiLSTMs, with no shared task linking them, don't spontaneously discover which verse word explains which commentary word.

```python
from IPython.display import display

@torch.no_grad()
def extract_bilstm_word_reps(verse_model, urai_model, rows, vocab, max_n=100):
    """Extract word-level hidden states from separate verse and urai BiLSTM models.
    Each occurrence also records which original (verse, urai) row it came from,
    so we can later tell whether a high-similarity word pair actually co-occurs
    in a real training example (same_verse) or only looks similar in embedding
    space across different examples (cross_verse)."""
    v_reps, u_reps = [], []

    verse_model.eval().to(DEVICE)
    for row_idx, row in enumerate(rows[:max_n]):
        toks  = [VERSE_TAG] + tokenize_words(row["verse"])
        ids   = encode_tokens(toks, vocab)
        ids_t = torch.tensor([ids], dtype=torch.long, device=DEVICE)
        lens  = torch.tensor([len(ids)], dtype=torch.long)
        _, enc = verse_model(ids_t, lens)
        for j in range(1, len(toks)):   # skip VERSE_TAG at index 0
            v_reps.append({"token": toks[j], "dataset": row["dataset"],
                           "vector": enc[0, j].cpu().numpy(), "row_idx": row_idx})
    verse_model.cpu()

    urai_model.eval().to(DEVICE)
    for row_idx, row in enumerate(rows[:max_n]):
        toks  = [URAI_TAG] + tokenize_words(row["urai"])
        ids   = encode_tokens(toks, vocab)
        ids_t = torch.tensor([ids], dtype=torch.long, device=DEVICE)
        lens  = torch.tensor([len(ids)], dtype=torch.long)
        _, enc = urai_model(ids_t, lens)
        for j in range(1, len(toks)):   # skip URAI_TAG at index 0
            u_reps.append({"token": toks[j], "dataset": row["dataset"],
                           "vector": enc[0, j].cpu().numpy(), "row_idx": row_idx})
    urai_model.cpu()

    return v_reps, u_reps

def avg_vectors(reps):
    """Average vectors per unique token, and also keep the set of row indices
    (original verse-urai examples) that token occurred in."""
    groups = defaultdict(list); ds_map = {}; pid_map = defaultdict(set)
    for r in reps:
        groups[r["token"]].append(r["vector"])
        ds_map[r["token"]] = r["dataset"]
        pid_map[r["token"]].add(r["row_idx"])
    return {t: (np.mean(vs, axis=0), ds_map[t], pid_map[t]) for t, vs in groups.items()}

TOP_K_BILSTM_PAIRS = 20

print("=== BiLSTM word-level cosine similarity (verse vs urai, separate models) ===")
bilstm_sims = {}
for nool in NOOLS + ["combined"]:
    nool_rows = get_nool_rows(nool, all_rows)
    v_reps, u_reps = extract_bilstm_word_reps(
        bilstm_verse[nool], bilstm_urai[nool], nool_rows, shared_vocab)

    v_avg = avg_vectors(v_reps); u_avg = avg_vectors(u_reps)
    v_words = list(v_avg.keys()); u_words = list(u_avg.keys())
    V_mat = np.stack([v_avg[w][0] for w in v_words])
    U_mat = np.stack([u_avg[w][0] for w in u_words])

    def norm_r(X): return X / (np.linalg.norm(X, axis=1, keepdims=True) + 1e-12)
    sim = norm_r(V_mat) @ norm_r(U_mat).T   # [|V|, |U|]

    flat = []
    for i in range(len(v_words)):
        vw = v_words[i]
        v_pids = v_avg[vw][2]
        for j in range(len(u_words)):
            uw = u_words[j]
            if vw == uw:
                continue
            u_pids = u_avg[uw][2]
            overlap = v_pids & u_pids
            flat.append({
                "i": i, "j": j,
                "score": float(sim[i, j]),
                "verse_word": vw,
                "urai_word": uw,
                "same_verse": len(overlap) > 0,
                "co_occurring_pairs": len(overlap),
            })
    flat.sort(key=lambda x: x["score"], reverse=True)

    # Greedy one-to-one selection: each verse word and urai word used at most once,
    # so "top 20" means 20 DISTINCT verse words paired with 20 DISTINCT urai words.
    top = []
    used_v, used_u = set(), set()
    for p in flat:
        if p["i"] in used_v or p["j"] in used_u:
            continue
        top.append(p)
        used_v.add(p["i"]); used_u.add(p["j"])
        if len(top) >= TOP_K_BILSTM_PAIRS:
            break

    n_same = sum(1 for p in top if p["same_verse"])
    bilstm_sims[nool] = {
        "v_reps": v_reps, "u_reps": u_reps,
        "top": top, "sim_mat": sim,
        "v_words": v_words, "u_words": u_words,
        "v_avg": v_avg, "u_avg": u_avg,
    }
    print(f"  [{nool:14s}]  sim mean={sim.mean():.3f}  max={sim.max():.3f}"
          f"  top: {top[0]['verse_word']} <-> {top[0]['urai_word']} ({top[0]['score']:.3f})")
    print(f"    top-{TOP_K_BILSTM_PAIRS}: {n_same} same-verse (co-occur in a real example), "
          f"{TOP_K_BILSTM_PAIRS - n_same} cross-verse (similar embeddings, never seen together)")
    display(pd.DataFrame(top)[["verse_word", "urai_word", "score", "same_verse", "co_occurring_pairs"]])

fig, axes = plt.subplots(1, len(NOOLS)+1, figsize=(4*(len(NOOLS)+1), 4), sharey=True)
for ax, nool in zip(axes, NOOLS + ["combined"]):
    sm = bilstm_sims[nool]["sim_mat"].flatten()
    ax.hist(sm, bins=40, color="tab:blue", alpha=0.75, edgecolor="white")
    ax.axvline(sm.mean(), color="red", linestyle="--", linewidth=1)
    ax.set_title(nool, fontsize=9); ax.set_xlabel("Cosine sim"); ax.grid(True, alpha=0.3)
axes[0].set_ylabel("Count")
plt.suptitle("BiLSTM: Pairwise verse-word vs urai-word cosine similarity\n(separate verse / urai encoders)", fontsize=11)
plt.tight_layout(); plt.show()
```

## Part B â€” Attention Encoders (separate verse + urai)

Same idea as Part A, but each side is now a small Transformer self-attention encoder (2 layers, 4 heads) instead of an LSTM â€” still trained separately per side with masked-language-modelling, still with no task connecting the two sides to each other.

This section is purely a visual clustering check (t-SNE below) â€” it doesn't run the same same-verse/cross-verse word-pair scoring as Part A.

```python
class AttnEncoder(nn.Module):
    """Transformer encoder for masked-LM pretraining; returns per-token contextual embeddings."""
    def __init__(self, vocab_size, d=128, nhead=4, n_layers=3, ff=256, dropout=0.1, max_len=512, pad_id=0):
        super().__init__()
        self.d = d
        self.embedding = nn.Embedding(vocab_size, d, padding_idx=pad_id)
        pe = torch.zeros(max_len, d)
        pos = torch.arange(max_len).unsqueeze(1).float()
        div = torch.exp(torch.arange(0, d, 2).float() * (-math.log(10000.0) / d))
        pe[:, 0::2] = torch.sin(pos * div); pe[:, 1::2] = torch.cos(pos * div)
        self.register_buffer("pe", pe.unsqueeze(0))
        self.drop = nn.Dropout(dropout)
        self.layers = nn.ModuleList([
            nn.TransformerEncoderLayer(d, nhead, dim_feedforward=ff, dropout=dropout,
                                       activation="gelu", batch_first=True, norm_first=True)
            for _ in range(n_layers)])
        self.norm    = nn.LayerNorm(d)
        self.lm_head = nn.Linear(d, vocab_size)
        self.out_dim = d

    def encode(self, ids, mask):
        x = self.drop(self.embedding(ids) * math.sqrt(self.d) + self.pe[:, :ids.size(1)])
        for layer in self.layers:
            x = layer(x, src_key_padding_mask=~mask)
        return self.norm(x)

    def forward(self, ids, mask):
        enc = self.encode(ids, mask)
        return self.lm_head(enc), enc
```

```python
def attn_mlm_collate(batch, vocab):
    pad_id = vocab[PAD]; mask_id = vocab[MASK]
    inputs, labels, lengths = [], [], []
    for ids in batch:
        x, y = mask_tokens(ids, mask_id)
        inputs.append(x); labels.append(y); lengths.append(len(x))
    mlen = max(lengths)
    inp_t = torch.full((len(batch), mlen), pad_id, dtype=torch.long)
    lab_t = torch.full((len(batch), mlen), -100,   dtype=torch.long)
    msk_t = torch.zeros(len(batch), mlen, dtype=torch.bool)
    for i, (x, y, l) in enumerate(zip(inputs, labels, lengths)):
        inp_t[i,:l] = torch.tensor(x); lab_t[i,:l] = torch.tensor(y); msk_t[i,:l] = True
    return inp_t, lab_t, msk_t

def train_attn(model, texts, vocab, prefix_toks, name, epochs=20, bs=32, lr=2e-3):
    ds     = MLMDataset(texts, vocab, prefix_toks)
    loader = DataLoader(ds, batch_size=bs, shuffle=True,
                        collate_fn=lambda b: attn_mlm_collate(b, vocab))
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-2)
    model.to(DEVICE); hist = []
    for ep in range(epochs):
        model.train(); total = n = 0
        for inp, lab, msk in loader:
            inp = inp.to(DEVICE); lab = lab.to(DEVICE); msk = msk.to(DEVICE)
            opt.zero_grad()
            logits, _ = model(inp, msk)
            loss = F.cross_entropy(logits.reshape(-1, logits.size(-1)),
                                   lab.reshape(-1), ignore_index=-100)
            loss.backward(); nn.utils.clip_grad_norm_(model.parameters(), 1.0); opt.step()
            total += loss.item() * inp.size(0); n += inp.size(0)
        hist.append(total / n)
    print(f"  [Attn {name:30s}]  final_loss={hist[-1]:.4f}")
    model.cpu(); return hist

attn_verse = {}; attn_urai = {}
print("=== Training Attention encoders (verse + urai per nool + combined) ===")
for nool in NOOLS + ["combined"]:
    nool_rows = get_nool_rows(nool, all_rows)
    verses = [r["verse"] for r in nool_rows]; urais = [r["urai"] for r in nool_rows]
    print(f"  [{nool}]  n={len(nool_rows)}")
    mv = AttnEncoder(len(shared_vocab), pad_id=shared_vocab[PAD])
    mu = AttnEncoder(len(shared_vocab), pad_id=shared_vocab[PAD])
    train_attn(mv, verses, shared_vocab, [VERSE_TAG], f"{nool}/verse", epochs=20)
    train_attn(mu, urais,  shared_vocab, [URAI_TAG],  f"{nool}/urai",  epochs=20)
    attn_verse[nool] = mv; attn_urai[nool] = mu
```

## Part C â€” Siamese Attention Networks

This is the one part of the notebook where the two sides are actually **trained together**. Starting from the Part B attention encoders, mean-pool each side's output into one vector, then train the whole thing so that cosine similarity between a matched verse+urai pair is pushed toward 1, and a random mismatched pair is pushed toward 0. One network per nool + one combined, 20 epochs each.

**Real result** â€” final training loss (MSE on cosine similarity, lower is better):

| Naaladiyar | Thirukadukam | Tholkappiyam | Combined |
|---|---|---|---|
| 0.0123 | 0.0168 | 0.0194 | 0.0123 |

All four converge to a very low loss â€” meaning the network *can* learn to separate real from fake pairs when it's actually given that as a training objective, unlike Parts A and B.

```python
class SiameseAttnNet(nn.Module):
    """Two AttnEncoders (verse + urai) with mean-pooling; cosine similarity between pooled reps."""
    def __init__(self, verse_enc, urai_enc):
        super().__init__()
        self.verse_enc = verse_enc; self.urai_enc = urai_enc

    def _pool(self, model, ids, mask):
        enc = model.encode(ids, mask)
        rep = (enc * mask.unsqueeze(-1).float()).sum(1) / mask.float().sum(1, keepdim=True).clamp(min=1)
        return F.normalize(rep, dim=-1)

    def get_verse_rep(self, ids, mask): return self._pool(self.verse_enc, ids, mask)
    def get_urai_rep(self,  ids, mask): return self._pool(self.urai_enc,  ids, mask)

    def forward(self, v_ids, v_msk, u_ids, u_msk):
        return F.cosine_similarity(self.get_verse_rep(v_ids, v_msk),
                                   self.get_urai_rep(u_ids,  u_msk))

class SiamesePairDataset(Dataset):
    def __init__(self, rows, vocab, neg_ratio=1):
        self.vocab = vocab; self.pairs = []
        n = len(rows)
        for i, row in enumerate(rows):
            self.pairs.append((row["verse"], row["urai"], 1.0))
            for _ in range(neg_ratio):
                j = i
                while j == i: j = random.randrange(n)
                self.pairs.append((row["verse"], rows[j]["urai"], 0.0))
    def __len__(self): return len(self.pairs)
    def __getitem__(self, i): return self.pairs[i]

def siamese_collate(batch, vocab):
    v_seqs, u_seqs, ys = [], [], []
    for verse, urai, y in batch:
        v_seqs.append(encode_tokens([VERSE_TAG]+tokenize_words(verse), vocab))
        u_seqs.append(encode_tokens([URAI_TAG] +tokenize_words(urai),  vocab))
        ys.append(y)
    def pad(seqs):
        mlen = max(len(s) for s in seqs)
        ids  = torch.full((len(seqs), mlen), vocab[PAD], dtype=torch.long)
        msk  = torch.zeros(len(seqs), mlen, dtype=torch.bool)
        for i, s in enumerate(seqs):
            ids[i,:len(s)] = torch.tensor(s); msk[i,:len(s)] = True
        return ids, msk
    vi, vm = pad(v_seqs); ui, um = pad(u_seqs)
    return vi, vm, ui, um, torch.tensor(ys, dtype=torch.float)
```

```python
def train_siamese(siamese, rows, vocab, name, epochs=20, bs=16, lr=1e-3):
    ds     = SiamesePairDataset(rows, vocab)
    loader = DataLoader(ds, batch_size=bs, shuffle=True,
                        collate_fn=lambda b: siamese_collate(b, vocab))
    opt = torch.optim.AdamW(siamese.parameters(), lr=lr, weight_decay=1e-2)
    siamese.to(DEVICE); hist = []
    for ep in range(epochs):
        siamese.train(); total = n = 0
        for vi, vm, ui, um, ys in loader:
            vi=vi.to(DEVICE); vm=vm.to(DEVICE); ui=ui.to(DEVICE); um=um.to(DEVICE); ys=ys.to(DEVICE)
            opt.zero_grad()
            loss = F.mse_loss(siamese(vi, vm, ui, um), ys)
            loss.backward(); nn.utils.clip_grad_norm_(siamese.parameters(), 1.0); opt.step()
            total += loss.item()*vi.size(0); n += vi.size(0)
        hist.append(total/n)
    print(f"  [Siamese {name:20s}]  final_loss={hist[-1]:.4f}")
    siamese.cpu(); return hist

siamese_models = {}
print("=== Training Siamese Attention models (3 per-nool + 1 combined) ===")
for nool in NOOLS + ["combined"]:
    nool_rows = get_nool_rows(nool, all_rows)
    s = SiameseAttnNet(attn_verse[nool], attn_urai[nool])
    train_siamese(s, nool_rows, shared_vocab, nool)
    siamese_models[nool] = s
```

## Siamese similarity analysis + t-SNE

For held-out verseâ€“urai pairs, score the *real* pairing (positive) and a randomly *mismatched* pairing (negative) with the trained Siamese network, and compare the two similarity distributions â€” plus a t-SNE of the resulting sentence vectors.

**Real result â€” positive similarity / negative similarity / separation:**

| | Naaladiyar | Thirukadukam | Tholkappiyam | Combined |
|---|---|---|---|---|
| positive (real pair) | 0.947 | 0.906 | 0.938 | 0.972 |
| negative (random pair) | 0.406 | 0.381 | 0.440 | 0.712 |
| separation | 0.540 | 0.525 | 0.498 | 0.260 |

**Positive result:** within any single text, the network clearly tells real pairs (~0.9+ similarity) apart from fake ones (~0.4 similarity) â€” a real signal, unlike Parts A/B.

**The issue:** once all texts are trained together ("combined"), separation nearly halves (0.540 â†’ 0.260) because the *negative* similarity jumps from ~0.4 to 0.712.

```python
@torch.no_grad()
def score_siamese(siamese, rows, vocab, max_n=80):
    siamese.eval().to(DEVICE)
    pos_s, neg_s = [], []
    subset = rows[:max_n]
    for i, row in enumerate(subset):
        j = i
        while j == i: j = random.randrange(len(subset))
        def enc(v, u):
            vi = torch.tensor([encode_tokens([VERSE_TAG]+tokenize_words(v), vocab)], dtype=torch.long, device=DEVICE)
            vm = torch.ones(1, vi.size(1), dtype=torch.bool, device=DEVICE)
            ui = torch.tensor([encode_tokens([URAI_TAG] +tokenize_words(u), vocab)], dtype=torch.long, device=DEVICE)
            um = torch.ones(1, ui.size(1), dtype=torch.bool, device=DEVICE)
            return float(siamese(vi, vm, ui, um).item())
        pos_s.append(enc(row["verse"], row["urai"]))
        neg_s.append(enc(row["verse"], subset[j]["urai"]))
    siamese.cpu()
    return pos_s, neg_s

print("=== Siamese similarity: positive vs negative pairs ===")
for nool, siamese in siamese_models.items():
    nool_rows = get_nool_rows(nool, all_rows)
    ps, ns = score_siamese(siamese, nool_rows, shared_vocab)
    print(f"  [{nool:14s}]  pos_mean={np.mean(ps):.3f}  neg_mean={np.mean(ns):.3f}  sep={np.mean(ps)-np.mean(ns):.3f}")

fig, axes = plt.subplots(1, len(siamese_models), figsize=(4*len(siamese_models), 4), sharey=True)
if len(siamese_models) == 1: axes = [axes]
for ax, (nool, siamese) in zip(axes, siamese_models.items()):
    ps, ns = score_siamese(siamese, get_nool_rows(nool, all_rows), shared_vocab, max_n=60)
    ax.hist(ps, bins=20, alpha=0.65, label="Positive", color="tab:blue")
    ax.hist(ns, bins=20, alpha=0.65, label="Negative", color="tab:red")
    ax.set_title(nool, fontsize=9); ax.set_xlabel("Cosine similarity"); ax.legend(fontsize=7)
    ax.grid(True, alpha=0.3)
plt.suptitle("Siamese Attention: Positive vs Negative Pair Distributions", fontsize=10)
plt.tight_layout(); plt.show()
```
